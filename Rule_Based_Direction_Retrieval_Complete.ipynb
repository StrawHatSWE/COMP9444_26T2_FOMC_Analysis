## Models

### Rule-Based

The implementation follows the paper's
[released rule code](https://github.com/gtfintechlab/fomc-hawkish-dovish/blob/main/code_model/rule_based.py):
dictionary terms are checked as lowercase substrings, Dovish conditions are
checked before Hawkish conditions, Panel C reverses direction, and unmatched
sentences are Neutral.

```python
examples = [
    "Inflation continued to decline.",
    "Inflation continued to rise.",
    "Inflation did not increase.",
    "The meeting ended at noon.",
]
for sentence in examples:
    prediction, matches = rule_based_predict_one(sentence)
    print(f"{LABEL_NAMES[prediction]:7s} | {sentence} | {matches}")
```

### Direction + Retrieval

The final customized model combines word and character TF-IDF with explicit
monetary-policy direction features. A cosine-neighbour memory is blended with
the three-class classifier. HDAR adds a hierarchy branch and is retained as an
ablation because its mean score is slightly lower.

```python
selected = json.loads(
    (ROOT / "results" / "selected_configs.json").read_text(encoding="utf-8")
)
pd.DataFrame(
    [
        {
            "seed": int(seed),
            "LR n-gram": str(config["tfidf_logistic_regression"]["word_ngram"]),
            "neighbours": config["direction_retrieval_augmented"][
                "neighbours"
            ],
            "retrieval alpha": config["direction_retrieval_augmented"][
                "retrieval_alpha"
            ],
            "neutral bias": config["direction_retrieval_augmented"][
                "neutral_bias"
            ],
        }
        for seed, config in selected.items()
    ]
)
```

## Experiment

Models are evaluated on raw test rows for seeds 5768, 78516, and 944601.
Weighted F1 is the primary metric; macro F1, accuracy, class F1, predictions,
and confusion matrices are also saved.

```python
aggregate = pd.read_csv(ROOT / "results" / "aggregate_results.csv")
columns = [
    "model_name",
    "weighted_f1_mean",
    "weighted_f1_std",
    "macro_f1_mean",
    "accuracy_mean",
    "dovish_f1_mean",
    "hawkish_f1_mean",
    "neutral_f1_mean",
]
aggregate[columns].round(4)
```

```python
rows = []
for seed in (5768, 78516, 944601):
    row = {"seed": seed}
    for label, model in [
        ("Direction + Retrieval", "direction_retrieval_augmented"),
        ("HDAR ablation", "hdar_ensemble_augmented"),
        ("Logistic Regression", "tfidf_logistic_regression"),
        ("Rule-Based", "rule_based"),
    ]:
        metrics = json.loads(
            (
                ROOT / "results" / model / f"seed_{seed}" / "metrics.json"
            ).read_text(encoding="utf-8")
        )
        row[f"{label} weighted F1"] = metrics["weighted_f1"]
        row[f"{label} macro F1"] = metrics["macro_f1"]
    rows.append(row)

per_seed = pd.DataFrame(rows)
per_seed["Direction + Retrieval - LR"] = (
    per_seed["Direction + Retrieval weighted F1"]
    - per_seed["Logistic Regression weighted F1"]
)
per_seed.round(4)
```

```python
display(
    Image(
        filename=str(ROOT / "results" / "weighted_f1_comparison.png"),
        width=900,
    )
)
```

## Discussion

```python
scores = aggregate.set_index("model_name")
final_model = scores.loc["direction_retrieval_augmented"]
hdar = scores.loc["hdar_ensemble_augmented"]
lr = aggregate.set_index("model_name").loc["tfidf_logistic_regression"]
rule = aggregate.set_index("model_name").loc["rule_based"]

seed_wins = int((per_seed["Direction + Retrieval - LR"] > 0).sum())
weighted_gap = final_model["weighted_f1_mean"] - lr["weighted_f1_mean"]
macro_gap = final_model["macro_f1_mean"] - lr["macro_f1_mean"]
bootstrap = pd.read_csv(
    ROOT / "results" / "bootstrap_direction_retrieval_vs_lr.csv"
)
display(bootstrap.round(4))
overall = bootstrap[bootstrap["evaluation"] == "three_seed_mean"].iloc[0]

display(
    Markdown(
        f"""Direction + Retrieval is the final customized model. Its point
estimate is higher than Logistic Regression on **{seed_wins}/3 seeds**, and
its mean weighted F1 is higher by **{weighted_gap:.4f}**. HDAR remains an
ablation because its mean weighted F1 is lower
({hdar['weighted_f1_mean']:.4f} versus
{final_model['weighted_f1_mean']:.4f}).

The grouped paired bootstrap gives a three-seed mean difference 95% CI of
**[{overall['ci_95_lower']:+.4f}, {overall['ci_95_upper']:+.4f}]**, with
**{overall['probability_final_gt_baseline']:.1%}** of resamples above Logistic
Regression. The individual intervals for seeds 5768 and 944601 cross zero, so
the improvement is supported in the mean but is not strong on every split.
Mean macro F1 is **{macro_gap:+.4f}** relative to Logistic Regression.

The Rule-Based mean weighted F1 is **{rule['weighted_f1_mean']:.4f}**, close to
the paper's reported **0.4966**."""
    )
)
```

## Error analysis

```python
errors = pd.read_csv(
    ROOT / "results" / "direction_retrieval_error_analysis.csv"
)
errors[
    [
        "sentence",
        "true_label",
        "predicted_label",
        "confidence",
        "error_category",
    ]
].head(10)
```

The main failures involve negation scope, mixed inflation and
employment directions, and implicit stance. Direction features and local
neighbours help, but the two near-zero seed-level bootstrap intervals show
that the gain is not uniformly large. The three seeds are alternative splits
of a small labelled dataset, so the mean result is not a general performance
guarantee.
